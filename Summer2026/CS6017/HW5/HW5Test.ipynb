```python
import numpy as np
import pandas as pd

# Load the features
features = pd.read_csv('UCIData/UCI HAR dataset/features.txt', header=None, names=['Feature', 'Description'])
features = features.set_index('Feature')

# Load the training data
X_train = pd.read_csv('UCIData/UCI HAR dataset/train/X_train.txt', header=None)
y_train = pd.read_csv('UCIData/UCI HAR dataset/train/y_train.txt', header=None)
subject_train = pd.read_csv('UCIData/UCI HAR dataset/train/subject_train.txt', header=None)

# Load the test data
X_test = pd.read_csv('UCIData/UCI HAR dataset/test/X_test.txt', header=None)
y_test = pd.read_csv('UCIData/UCI HAR dataset/test/y_test.txt', header=None)
subject_test = pd.read_csv('UCIData/UCI HAR dataset/test/subject_test.txt', header=None)

# Combine train and test data
X = pd.concat([X_train, X_test], axis=0)
y = pd.concat([y_train, y_test], axis=0)
subject = pd.concat([subject_train, subject_test], axis=0)

# Reset index
X = X.reset_index(drop=True)
y = y.reset_index(drop=True)
subject = subject.reset_index(drop=True)

# Combine into a single DataFrame
data = pd.concat([X, y, subject], axis=1)
data.columns = [f'feature_{i}' for i in range(X.shape[1])] + ['y', 'subject']

# Display the first few rows
data.head()
```
