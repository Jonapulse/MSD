```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.feature_selection import f_classif, SelectPercentile

moneyDF = pd.read_csv("data_banknote_authentication.csv", header=None)
moneyDF.columns = ['variance', 'kewness', 'curtosis', 'entropy', 'class']

sns.pairplot(moneyDF, hue='class')
```

```python

predictor_features = moneyDF.columns[:-1]
selector = SelectPercentile(f_classif)
X_new = selector.fit_transform(moneyDF[predictor_features].values, moneyDF["label"].values)
print(predictor_features)
print("P Values:", selector.pvalues_)
for pr in zip(predictor_features, selector.pvalues_):
    print(pr[0], pr[1])
```
